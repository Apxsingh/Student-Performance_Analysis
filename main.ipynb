## Student Performance Analysis
### Data Preprocessing & Visualization

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Step 1: Load Dataset
print("Loading Dataset...\n")

df = pd.read_csv("your_csv_file_directory")

# Display first 5 rows
print("First 5 Rows:")
print(df.head())
```

```python
# Step 2: Basic Information

print("\nDataset Information:")
print(df.info())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

```python
# Step 3: Statistical Summary

print("\nStatistical Summary:")
print(df.describe())
```

```python
# Step 4: Check Missing Values

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Step 5: Remove Duplicate Rows

duplicates = df.duplicated().sum()
print(f"\nDuplicate Rows: {duplicates}")

df.drop_duplicates(inplace=True)
```

```python
# Step 6: Add Average Marks Column

df["AverageMarks"] = (
    df["Math"] +
    df["Science"] +
    df["English"]
) / 3

print("\nDataset with Average Marks:")
print(df.head())
```

```python
# Step 7: Find Top Performer

top_student = df.loc[df["AverageMarks"].idxmax()]

print("\nTop Performer:")
print(top_student)
```

```python
# Step 8: Visualization

# Bar Chart
plt.figure(figsize=(10, 5))

plt.bar(df["Name"], df["Math"])

plt.title("Math Scores of Students")
plt.xlabel("Students")
plt.ylabel("Math Marks")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# Line Graph
plt.figure(figsize=(10, 5))

plt.plot(df["Name"], df["Attendance"], marker='o')

plt.title("Attendance of Students")
plt.xlabel("Students")
plt.ylabel("Attendance Percentage")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# Histogram
plt.figure(figsize=(8, 5))

plt.hist(df["Science"], bins=5)

plt.title("Science Marks Distribution")
plt.xlabel("Science Marks")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

# Pie Chart
study_hours = df["StudyHours"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    study_hours,
    labels=study_hours.index,
    autopct='%1.1f%%'
)

plt.title("Study Hours Distribution")

plt.show()
```

```python
# Step 9: Insights

print("\n Insights ")

high_attendance = df[df["Attendance"] > 90]

print(
    f"Students with attendance above 90%: "
    f"{len(high_attendance)}"
)

avg_study = df["StudyHours"].mean()

print(f"Average Study Hours: {avg_study:.2f}")

avg_marks = df["AverageMarks"].mean()

print(f"Average Marks of Class: {avg_marks:.2f}")

print("\nProject Completed Successfully!")
```
